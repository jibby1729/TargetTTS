# Test Preprocessing
Load LibriSpeech samples, play the audio, and view raw vs normalized transcripts.

```python
import sys
import os

# Set working directory to project root so relative paths in manifests work
os.chdir(os.path.join(os.path.dirname(os.path.abspath("__file__")), ".."))
sys.path.insert(0, ".")

import json
import IPython.display as ipd
from src.preprocess import preprocess_audio, normalize_text, TARGET_SAMPLE_RATE
```

```python
# Load manifest
manifest_path = "data/librispeech/manifests/test-clean.jsonl"

with open(manifest_path, "r", encoding="utf-8") as f:
    entries = [json.loads(line) for line in f]

print(f"Loaded {len(entries)} utterances")
```

```python
# Pick a sample (change index to browse different utterances)
idx = 0
entry = entries[idx]

print(f"File:       {entry['audio_path']}")
print(f"Speaker:    {entry['speaker_id']}")
print(f"Duration:   {entry['duration_s']:.2f}s")
```

```python
# Load and preprocess audio
waveform = preprocess_audio(entry["audio_path"])

print(f"Shape:      {list(waveform.shape)}")
print(f"Amp range:  [{waveform.min():.4f}, {waveform.max():.4f}]")

# Play the audio
ipd.Audio(waveform.squeeze().numpy(), rate=TARGET_SAMPLE_RATE)
```

```python
# Compare raw vs normalized transcript
raw = entry["transcript"]
normalized = normalize_text(raw)

print(f"Raw:        {raw}")
print(f"Normalized: {normalized}")

if raw == normalized:
    print("\n(No change — LibriSpeech ground truth is already clean)")
else:
    print(f"\nDifferences detected!")
```

```python
# Browse several samples at once
for i in range(5):
    e = entries[i]
    w = preprocess_audio(e["audio_path"])
    raw = e["transcript"]
    norm = normalize_text(raw)

    print(f"[{i}] Speaker {e['speaker_id']} | {e['duration_s']:.1f}s")
    print(f"    Raw:  {raw}")
    print(f"    Norm: {norm}")
    display(ipd.Audio(w.squeeze().numpy(), rate=TARGET_SAMPLE_RATE))
    print()
```
